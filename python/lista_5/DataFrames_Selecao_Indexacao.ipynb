```python
import pandas as pd

#Concatenar DataFrames
df1 = pd.DataFrame(
    { "A": ["A0", "A1", "A2", "A3"],
      "B": ["B0", "B1", "B2", "B3"],
      "C": ["C0", "C1", "C2", "C3"],
      "D": ["D0", "D1", "D2", "D3"],
      },
    )
df2 = pd.DataFrame(
    { "A": ["A4", "A5", "A6", "A7"],
      "B": ["B4", "B5", "B6", "B7"],
      "C": ["C4", "C5", "C6", "C7"],
      "D": ["D4", "D5", "D6", "D7"],
      },
    )

print(df1)
print("-"*30)
print(df2)
```

```python
print("Concatenar indice desorganizado")
df = pd.concat([df1, df2])
print(df)
print("-"*30)

print("Concatenar indice organizado")
dfIndex = pd.concat([df1, df2], ignore_index = True)
print(dfIndex)
```

```python
# Acessando e imprimindo um valor específico usando iloc
print(dfIndex)

print("Acessando uma coluna específica:")
print(dfIndex.iloc[5])
print('-' * 30)  # Linha separadora

print("Acessando uma coluna e linha específica:")
print("Posição", 1, 3, dfIndex.iloc[1, 3])  # Acessa o valor da primeira linha e segunda coluna
print("Posição", 5, 3, dfIndex.iloc[5, 3])  # Acessa o valor da primeira linha e segunda coluna
```

```python
left2 = pd.DataFrame(
    {"A": ["A0", "A1", "A2"], "B": ["B0", "B1", "B2"]}, index=["K0", "K1", "K2"]
)

right2 = pd.DataFrame(
    {"C": ["C0", "C2", "C3"], "D": ["D0", "D2", "D3"]}, index=["K0", "K2", "K3"]
)

right2.join(left2)
```

```python
right2.join(left2, how = "inner")
```

```python
pd.merge(left2, right2, left_index = True, right_index = True, how = "inner")
```

Por fim, podemos resolver um problema com dados reais. Abaixo importamos dados de preços de uma ação da Ibovespa e data frames separados. Queremos juntar os dois data frames.

No caso, ambos data frames possuem o índice em comum, que neste caso, possuem os valores da data, referentes ao dia de cada preço da ação. Escolhemos juntar por meio a função merge(), identificando os índices com left_index e right_index e a forma de junção como "inner".

```python
# Data com yahoo finance

from pandas_datareader import data as pdr
!pip install -q U yfinance
import yfinance as yf
yf.pdr_override()

# Busca os preços ajustados
prices = pdr.get_data_yahoo(['ITUB4.SA'])['Adj Close']

bvsp = pdr.get_data_yahoo('^BVSP')[['Adj Close']]

# Junta os dados
pd.merge(prices, bvsp, left_index = True, right_index = True, how = 'inner')
```
